```python
# Cat and Dog Image Classification using SVM

import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

from skimage.feature import hog
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# Step 1: Dataset Path

# Folder structure:
# PetImages/
#    ├── Cat/
#    └── Dog/

dataset_path = "PetImages"

# Resize all images to this size
IMG_SIZE = (128, 128)

# Lists to store features and labels
images = []
labels = []

# Step 2: Load All Images and Extract HOG Features

for label, folder in enumerate(["Cat", "Dog"]):

    folder_path = os.path.join(dataset_path, folder)

    for file in os.listdir(folder_path):

        img_path = os.path.join(folder_path, file)

        try:
            # Read image
            img = cv2.imread(img_path)

            # Skip corrupted images
            if img is None:
                continue

            # Convert to grayscale
            img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

            # Resize image
            img = cv2.resize(img, IMG_SIZE)

            # Extract HOG Features
            feature = hog(
                img,
                orientations=9,
                pixels_per_cell=(8, 8),
                cells_per_block=(2, 2),
                block_norm="L2-Hys"
            )

            images.append(feature)
            labels.append(label)

        except:
            continue

print("Total Images Loaded:", len(images))

# Step 3: Prepare Dataset

X = np.array(images)
y = np.array(labels)

# Step 4: Split Dataset into Training and Testing

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Step 5: Train Support Vector Machine (SVM)

model = SVC(kernel="rbf", C=10, gamma="scale")
model.fit(X_train, y_train)

# Step 6: Predict on Train and Test Data

train_pred = model.predict(X_train)
test_pred = model.predict(X_test)

# Step 7: Display Accuracy

train_accuracy = accuracy_score(y_train, train_pred)
test_accuracy = accuracy_score(y_test, test_pred)

print("\nTraining Accuracy :", round(train_accuracy * 100, 2), "%")
print("Testing Accuracy  :", round(test_accuracy * 100, 2), "%")

# Step 8: Display Classification Report

print("\nClassification Report:\n")

print(classification_report(
    y_test,
    test_pred,
    target_names=["Cat", "Dog"]
))
```

```python
# Predict a Single Image using Trained Model

# Give the path of any image
test_image = "PetImages/Cat/1000.jpg"
# Example:
# test_image = "PetImages/Dog/2500.jpg"

img = cv2.imread(test_image)

if img is None:
    print("Image not found!")

else:
    # Display original image
    display_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Preprocess image
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    gray = cv2.resize(gray, IMG_SIZE)

    # Extract HOG features
    feature = hog(
        gray,
        orientations=9,
        pixels_per_cell=(8, 8),
        cells_per_block=(2, 2),
        block_norm="L2-Hys"
    )

    # Predict class
    prediction = model.predict([feature])[0]

    predicted_class = "Cat" if prediction == 0 else "Dog"

    # Show image with prediction
    plt.imshow(display_img)
    plt.title("Prediction: " + predicted_class)
    plt.axis("off")
    plt.show()

    print("Predicted Class:", predicted_class)
```
