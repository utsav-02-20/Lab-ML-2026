```python
# Data
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Split data
from sklearn.model_selection import train_test_split

#model train
from sklearn.linear_model import LinearRegression

#model Evaluation
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

#randon forest regressor

from sklearn.ensemble import RandomForestRegressor
```

```python
data = pd.read_csv('USA_Housing - USA_Housing.csv')
data.head()
```

```python
# target value is Price

# variables are avg home age , avg no of room , avg no of bedsrooms , area population , address 
```

```python
data.columns
```

```python
dependent_list = ["Price"]

independent_list = [
    "Avg. Area Income",
    "Avg. Area House Age",
    "Avg. Area Number of Rooms",
    "Avg. Area Number of Bedrooms",
    "Area Population"
]

for col in independent_list:
    plt.figure(figsize=(7, 4))
    
    plt.scatter(data[col], data["Price"])
    
    plt.xlabel(col)
    plt.ylabel("Price")
    plt.title(f"{col} vs Price")
    
    plt.show()
```

```python
# heat map of dependence 
plt.figure(figsize=(10, 7))

sns.heatmap(
    data[independent_list + ["Price"]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
x = data[independent_list]

y = data["Price"]

# split data 
```

```python
# train test split 
x_train , x_test , y_train , y_test = train_test_split(
    x , y , 
    test_size=0.2 , random_state=42
)
```

```python
# using linear regration 
model = LinearRegression()

model.fit(x_train , y_train)
```

```python
# test model on data which are not seen 
y_pred = model.predict(x_test)
```

```python
# for regression 

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("R2 Score:", r2)
```

```python
# visualize actual vss predicted 
plt.figure(figsize=(7, 5))

plt.scatter(y_test, y_pred)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted House Prices")

plt.show()
```

```python
# create a model for random forest regressor 

# basic configuration
# rf_model = RandomForestRegressor(
#     n_estimators=100,
#     random_state=42
# )

# stronger configuration
rf_model = RandomForestRegressor(
    n_estimators=30000,
    max_depth=None,
    min_samples_split=2,
    random_state=42,
    n_jobs=-1
)
```

```python
# train that model on our existing data set 
rf_model.fit(x_train, y_train)
```

```python
# predict values on our test data set 
rf_pred = rf_model.predict(x_test)
```

```python
# evaluate the random forest model 
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_mse = mean_squared_error(y_test, rf_pred)
rf_rmse = np.sqrt(rf_mse)
rf_r2 = r2_score(y_test, rf_pred)

print("MAE:", rf_mae)
print("MSE:", rf_mse)
print("RMSE:", rf_rmse)
print("R2 Score:", rf_r2)
```

```python
# actual vs predicted for rf 
plt.figure(figsize=(7, 5))

plt.scatter(y_test, rf_pred)

# Perfect prediction line
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Random Forest: Actual vs Predicted Prices")

plt.show()
```
