```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load dataset
df = pd.read_csv("Mall_Customers.csv")

# Select three features
X = df[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']]

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA to reduce dimensions from 3 to 2
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Create a DataFrame with reduced features
pca_df = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])

# Print reduced features
print("Reduced Features:")
print(pca_df.head())

# Print explained variance ratio
print("\nExplained Variance Ratio:")
print("PC1:", pca.explained_variance_ratio_[0])
print("PC2:", pca.explained_variance_ratio_[1])
print("Total Explained Variance:",
      pca.explained_variance_ratio_.sum())

# Plot explained variance ratio
plt.figure(figsize=(7, 5))
plt.bar(['PC1', 'PC2'], pca.explained_variance_ratio_)
plt.ylabel('Explained Variance Ratio')
plt.title('PCA Explained Variance Ratio')
plt.show()

# Plot 2D scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(pca_df['PC1'], pca_df['PC2'], s=50)
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('PCA: Customer Data in 2D')
plt.grid(True)
plt.show()
```
